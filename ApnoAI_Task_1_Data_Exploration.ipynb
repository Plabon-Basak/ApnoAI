## 2. Data overview

```python
df.info()
```

**Observation:** There are 50 intern records with a mix of numerical and categorical columns.

```python
df.isnull().sum()
```

**Observation:** There are no missing values, so no cleaning is needed before analysis.

```python
df.describe()
```

**Observation:** On average, interns were active for 16.82 days, completed 1.8 tasks, scored 69 on the quiz, practiced 79.08 questions and studied 6.98 hours per week. Quiz scores range from 32 to 96 and weekly hours from 2 to 14, so participation levels differ a lot between interns.

## 3. Track distribution

```python
print(df['track'].value_counts())

df['track'].value_counts().plot(kind='bar', title='Interns per Track', xlabel='Track', ylabel='Number of Interns', rot=0)
plt.show()
```

**Observation:** Web Development and Data Analytics have 20 interns each, while AI / ML has only 10. Comparisons involving AI / ML are based on a smaller group.

## 4. Quiz score distribution

```python
df['quiz_score'].plot(kind='hist', bins=10, title='Quiz Score Distribution', edgecolor='white')
plt.xlabel('Quiz Score')
plt.ylabel('Number of Interns')
plt.show()
```

**Insight:** Quiz scores are spread fairly evenly from 32 to 96 with no single dominant peak. About a third of interns score below 60, while about half score 70 or above.

## 5. Average quiz score by track

```python
print(df.groupby('track')['quiz_score'].mean().round(2))

plt.figure(figsize=(8,4))
sns.barplot(data=df, x='track', y='quiz_score', errorbar=None)
plt.title('Average Quiz Score by Track')
plt.xlabel('Track')
plt.ylabel('Average Quiz Score')
plt.show()
```

**Insight:** Average quiz scores are almost identical across tracks (AI / ML 68.3, Data Analytics 68.9, Web Development 69.5). Track does not seem to affect performance, so differences between interns come from other factors such as effort.

## 6. Hours per week vs quiz score

```python
plt.figure(figsize=(9,4))
sns.scatterplot(data=df, x='hours_per_week', y='quiz_score', hue='track')
plt.title('Hours per Week vs Quiz Score')
plt.xlabel('Hours per Week')
plt.ylabel('Quiz Score')
plt.show()
```

**Insight:** Interns who study more hours per week score higher, and this pattern holds in all three tracks.

## 7. Satisfaction

```python
print(df['satisfied'].value_counts())

sns.countplot(data=df, x='satisfied')
plt.title('Satisfied Interns Count')
plt.xlabel('Satisfied')
plt.ylabel('Number of Interns')
plt.show()
```

**Observation:** 30 interns (60%) are satisfied and 20 (40%) are not.

```python
df.groupby('satisfied')[['quiz_score','hours_per_week','tasks_completed','days_active']].mean().round(2)
```

**Insight:** Satisfied interns are far more active. They average a quiz score of 81.4, 9.3 hours per week and 2.5 tasks, compared with 50.4, 3.6 hours and 0.75 tasks for unsatisfied interns. Low engagement and low satisfaction go together.

## 8. Task completion and quiz performance (6th insight)

```python
df.groupby('tasks_completed')['quiz_score'].mean().round(2)
```

```python
df.groupby('tasks_completed')['quiz_score'].mean().plot(
    kind='bar',
    title='Average Quiz Score by Tasks Completed',
    xlabel='Tasks Completed',
    ylabel='Average Quiz Score',
    rot=0
)
plt.show()
```

**Insight:** Interns who completed more tasks scored higher. The average quiz score rises from about 36 with 0 tasks to about 90 with 3 tasks.

# Study time and quiz performance

```python
high = df[df['hours_per_week'] >= 8]['quiz_score'].mean()
low = df[df['hours_per_week'] < 8]['quiz_score'].mean()
print(f"8+ hours/week: {high:.1f}")
print(f"Under 8 hours/week: {low:.1f}")
```

**Insight:** Interns who study 8 or more hours per week (20 interns) average a quiz score of 86.8, compared with 57.1 for those who study fewer hours.

# Correlation between numeric columns

```python
num_cols = ['days_active','tasks_completed','quiz_score','practice_questions','hours_per_week']
plt.figure(figsize=(7,5))
sns.heatmap(df[num_cols].corr(), annot=True, cmap='Blues', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

**Insight:** All numeric columns are strongly correlated with each other (0.93 to 0.99). Interns who are active more days also practice more, study longer, complete more tasks and score higher. This shows a link, not proof of cause: engaged interns simply do more of everything.

## Conclusion

- The data is clean: 50 records, no missing values and no duplicates.
- Track makes almost no difference to quiz scores. Engagement matters much more.
- More study hours, more tasks and more practice go with higher quiz scores.
- Satisfied interns are the highly engaged ones, so getting low-activity interns to participate more could improve both performance and satisfaction.
- The dataset is small (50 interns, only 10 in AI / ML), so these patterns should be checked with more data.
